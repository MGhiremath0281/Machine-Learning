# Understanding List Comprehension in Python

A simple trick to understand **list comprehension** in Python is to break it down into three parts:

1. **Expression**: What you want to do with each item.
2. **Iteration**: The loop that goes through the items.
3. **Condition (Optional)**: A filter to include only certain items.

## Basic Syntax:
```python
[expression for item in iterable if condition]

````python
#Function to transpose a 3x3 matrix.
def transpose_matrix(matrix):
    return [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]

# Original matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Transposing the matrix
transposed = transpose_matrix(matrix)

# Printing matrices
print("Original Matrix:")
for row in matrix:
    print(row)

print("Transposed Matrix:")
for row in transposed:
    print(row)
````

````python
#Function to flatten a nested list.
def flatten_list(nested_list):
    return [item for sublist in nested_list for item in sublist]

# Nested list
nested_list = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]

# Flattening the list
flattened = flatten_list(nested_list)

# Printing lists
print("Original Nested List:", nested_list)
print("Flattened List:", flattened)
````

````python
#Function to rotate a list by n positions.
def rotate_list(lst, n):
    n = n % len(lst)  # To handle cases where n > len(lst)
    return lst[n:] + lst[:n]

# Original list
numbers = [1, 2, 3, 4, 5]

# Rotating by 2 positions
rotated_list = rotate_list(numbers, 2)

# Printing lists
print("Original List:", numbers)
print("Rotated List:", rotated_list)
````

````python
#Function to find common elements between two lists.
def list_intersection(lst1, lst2):
    return list(set(lst1) & set(lst2))

# Two lists
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]

# Finding intersection
intersection = list_intersection(list1, list2)

# Printing result
print("Intersection:", intersection)
````

````python
# Create the set
fruits = {"apple", "banana", "cherry"}

# Print the set
print("Initial set:", fruits)
````

````python
# Create a second set
citrus = {"orange", "lemon", "lime"}

# Union: elements in either fruits or citrus
union_set = fruits.union(citrus)
print("Union of fruits and citrus:", union_set)

# Intersection: common elements in both sets
intersection_set = fruits.intersection(citrus)
print("Intersection of fruits and citrus:", intersection_set)

# Difference: elements in fruits that are not in citrus
difference_set = fruits.difference(citrus)
print("Difference (fruits - citrus):", difference_set)

# Symmetric Difference: elements in either set but not both
sym_diff_set = fruits.symmetric_difference(citrus)
print("Symmetric Difference:", sym_diff_set)
````

````python
# Define two sample dictionaries
dict1 = {"a": 1, "b": 2, "c": 3}
dict2 = {"b": 20, "d": 4, "e": 5}

print("Original Dictionaries:")
print("dict1:", dict1)
print("dict2:", dict2)

# 1. Using update() (Modifies dict1)
merged_update = dict1.copy()  # To keep dict1 unchanged
merged_update.update(dict2)
print("\nMerged using update():", merged_update)

# 2. Using | Operator (Python 3.9+)
merged_pipe = dict1 | dict2
print("\nMerged using | operator:", merged_pipe)

# 3. Using Dictionary Unpacking (**)
merged_unpack = {**dict1, **dict2}
print("\nMerged using ** unpacking:", merged_unpack)

# 4. Using Dictionary Comprehension
merged_comprehension = {key: value for d in (dict1, dict2) for key, value in d.items()}
print("\nMerged using dictionary comprehension:", merged_comprehension)

# 5. Handling Key Conflicts (Merging Without Overwriting)
merged_no_overwrite = {key: dict1.get(key, dict2.get(key)) for key in dict1.keys() | dict2.keys()}
print("\nMerged without overwriting keys:", merged_no_overwrite)
````

```Write a function is_prime(n) that checks if n is a prime number.```

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))  # Output: True
print(is_prime(10))  # Output: False
````

````python
def fibonacci(n):
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

print(fibonacci(5))  # Output: [0, 1, 1, 2, 3]
````
