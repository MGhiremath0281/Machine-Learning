```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
encoded_data = pd.get_dummies(df[['gender', 'race/ethnicity', 'lunch', 'test preparation course']], drop_first=True)
```

```python
from sklearn.preprocessing import OrdinalEncoder

# Correct the categories to match the actual values in the dataframe
# and include all unique values
education_order = [['some high school', 'high school', 'some college', "associate's degree", "bachelor's degree", "master's degree"]]
encoder = OrdinalEncoder(categories=education_order)
df['education_level_encoded'] = encoder.fit_transform(df[['parental level of education']])
```

```python
df = df.drop(['gender', 'race/ethnicity', 'lunch', 'test preparation course','education_level_encoded'], axis=1)
```

```python
df
```

```python
data = pd.concat([df, encoded_data], axis=1)
```

```python
data = data.applymap(lambda x: int(x) if isinstance(x, bool) else x)
```
