````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
````

````python
# Creating a synthetic dataset based on World Happiness Report structure
np.random.seed(42)

countries = [
    "Norway", "Denmark", "Switzerland", "Finland", "Netherlands",
    "Canada", "New Zealand", "Sweden", "Australia", "United Kingdom",
    "United States", "Germany", "Belgium", "France", "Japan",
    "India", "China", "Brazil", "South Africa", "Russia"
]

regions = [
    "Western Europe", "Western Europe", "Western Europe", "Western Europe", "Western Europe",
    "North America", "Australia and New Zealand", "Western Europe", "Australia and New Zealand", "Western Europe",
    "North America", "Western Europe", "Western Europe", "Western Europe", "East Asia",
    "South Asia", "East Asia", "Latin America", "Sub-Saharan Africa", "Eastern Europe"
]

# Generate random values for each metric within a reasonable range
happiness_scores = np.round(np.random.uniform(3, 8, len(countries)), 2)
gdp_per_capita = np.round(np.random.uniform(0.5, 1.5, len(countries)), 2)
social_support = np.round(np.random.uniform(0.5, 1.5, len(countries)), 2)
healthy_life_expectancy = np.round(np.random.uniform(50, 85, len(countries)), 2)
freedom = np.round(np.random.uniform(0.2, 1.0, len(countries)), 2)
generosity = np.round(np.random.uniform(0.1, 0.5, len(countries)), 2)
corruption = np.round(np.random.uniform(0.1, 0.9, len(countries)), 2)

# Create DataFrame
happiness_df = pd.DataFrame({
    "Country": countries,
    "Region": regions,
    "Happiness Score": happiness_scores,
    "GDP per Capita": gdp_per_capita,
    "Social Support": social_support,
    "Healthy Life Expectancy": healthy_life_expectancy,
    "Freedom to Make Life Choices": freedom,
    "Generosity": generosity,
    "Perceptions of Corruption": corruption
})

# Display first few rows
happiness_df.head()
````

```Set style```

````python
sns.set_style("whitegrid")
````

````python
happiness_df.info()
````

```1.Happiness score visulization```

````python
plt.figure(figsize=(12,6))
sns.histplot(happiness_df['Happiness Score'],bins=10,kde=True,color='skyblue')
plt.title('Distribution of Happiness Scores')
plt.xlabel('Happiness Score')
plt.ylabel('Frequency')
plt.show()
````

````python
plt.figure(figsize=(12,6))
sns.boxplot(y='Happiness Score', color='lightcoral', data=happiness_df) # Add data=happiness_df
plt.title('Boxplot of Happiness Scores')
plt.ylabel('Happiness Score')
plt.show()
````

``` 2. GDP per Capita Visualization```

````python
plt.figure(figsize=(12,6))
sns.boxplot(y='GDP per Capita',color='lightcoral',data=happiness_df)
plt.title('Boxplot of GDP per Capita')
plt.ylabel('GDP per Capita')
plt.show()
````

````python
plt.figure(figsize=(12,6))
sns.scatterplot(x='GDP per Capita',color='red',y='Happiness Score',data=happiness_df)
plt.title('Scatterplot of GDP per Capita vs Happiness Score')
plt.xlabel('GDP per Capita')
plt.ylabel('Happiness Score')
plt.show()
````

```Social support visulaliztion```

````python
plt.figure(figsize=(12,6))
sns.violinplot(x='Social Support',color='gold',data=happiness_df)
plt.title('Boxplot of Social Support')
plt.xlabel('Social Support')
plt.show()
````

````python
plt.figure(figsize=(12,6))
sns.regplot(x='Social Support',y='Happiness Score',data=happiness_df,color='green')
plt.title('Scatterplot of Social Support vs Happiness Score')
plt.xlabel('Social Support')
plt.ylabel('Happiness Score')
````

```4. Healthy Life Expectancy Visualization ```

````python
plt.figure(figsize=(12, 6))
sns.kdeplot(happiness_df['Healthy Life Expectancy'], shade=True, color='green')
plt.title("Healthy Life Expectancy Density Plot")
plt.xlabel("Healthy Life Expectancy")
plt.ylabel("Density")
plt.show()
````

````python
plt.figure(figsize=(12, 6))
sns.scatterplot(x=happiness_df['Healthy Life Expectancy'], y=happiness_df['Happiness Score'], color='teal')
plt.title("Healthy Life Expectancy vs Happiness Score")
plt.xlabel("Healthy Life Expectancy")
plt.ylabel("Happiness Score")
plt.show()
````

``` 5. Freedom to Make Life Choices Visualization```

````python
plt.figure(figsize=(12,6))
sns.stripplot(x='Freedom to Make Life Choices',data=happiness_df,jitter=True,color='purple')
plt.title('Stripplot of Freedom to Make Life Choices')
plt.xlabel('Freedom to Make Life Choices')
````

````python
plt.figure(figsize=(12,6))
sns.regplot(x='Freedom to Make Life Choices',y='Happiness Score',data=happiness_df,color='orange')
plt.title('Scatterplot of Freedom to Make Life Choices vs Happiness Score')
plt.xlabel('Freedom to Make Life Choices')
plt.ylabel('Happiness Score')
````

```6. Generosity Visualization```

````python
plt.figure(figsize=(12,6))
sns.histplot(happiness_df['Generosity'],bins=10,kde=True,color='purple')
plt.title('Distribution of Generosity')
plt.xlabel('Generosity')
plt.ylabel('Frequency')
plt.show()
````

```7. Perceptions of Corruption Visualization```

````python
plt.figure(figsize=(12,6))
sns.barplot(x='Perceptions of Corruption',y='Country',data=happiness_df,color='#ef5675')
plt.title('Barplot of Perceptions of Corruption')
plt.xlabel('Perceptions of Corruption')
plt.show()
````

```8. Pairplot to see relationships between numerical features```

````python
sns.pairplot(happiness_df,diag_kind='kde',palette='husl')
plt.suptitle('Pairplot of Numerical Features')
plt.show()
````

```9. Correlation Heatmap```

````python
# Select only numerical features for correlation calculation
numerical_features = happiness_df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 6))
sns.heatmap(numerical_features.corr(), annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
````

```10. Bubble Chart (Size of bubbles represents GDP per Capita)```

````python
plt.figure(figsize=(12, 6))
plt.scatter(happiness_df['Happiness Score'], happiness_df['Healthy Life Expectancy'],
            s=happiness_df['GDP per Capita']*300, alpha=0.5, c=happiness_df['Happiness Score'], cmap='viridis')
plt.colorbar(label='Happiness Score')
plt.title('Bubble Chart of Happiness vs Healthy Life Expectancy')
plt.xlabel('Happiness Score')
plt.ylabel('Healthy Life Expectancy')
plt.show()
````
